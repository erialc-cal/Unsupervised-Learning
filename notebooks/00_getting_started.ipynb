### Option 1 — Google Colab

Colab already provides many scientific Python packages. Run the import cell below first; do not reinstall packages merely because a tutorial contains an installation command. ### Option 2 — Miniconda or Anaconda

Conda distributions also include or can safely install many scientific packages. Create the dedicated environment described by the repository's `environment.yml`; avoid modifying the shared `base` environment. From the repository root, in the terminal (ie you should see the left handle showing `USER@Macbook-USER % FOLDER_NAME` on Mac, or `PS C:\Users\USER\...\FOLDER_NAME` on Windows) run:

```bash
conda env create -f environment.yml
conda activate statgr4244
python -m jupyter lab
```

To update an environment after the course file changes, run `conda env update -f environment.yml --prune`. Prefer Conda packages within this environment; use `pip` only when a required package is unavailable from the configured Conda channel.



### Option 3 — Plain Python

Create a virtual environment in the repository rather than installing globally. On macOS or Linux, and from where you want to put your repository in the terminal, run:

```bash
python3 -m venv .venv
source .venv/bin/activate
python -m pip install --upgrade pip
python -m pip install -r requirements.txt
python -m jupyter lab
```

On Windows PowerShell, activation is `.\.venv\Scripts\Activate.ps1`; the remaining `python -m ...` commands are the same. When returning to the course later, activate the same environment instead of creating it again. Run `deactivate` when finished.
